# 10. Fazit und Ausblick

Dieses Notebook enthält keinen Code. Es beantwortet die Fragestellungen aus `00_projekt.ipynb`
mit den Ergebnissen der Notebooks 03 bis 08 und beschreibt Grenzen, Ausblick und Reflexion.

## Beantwortung der Fragestellungen

**1. Nationale Prägung und Gruppen** (`03_reduktion`, `04_communities`). Das Netz ist stark
national geprägt. 43.1 % der Kanten im Arbeitsnetz verbinden Landsleute, bei zufällig verbundenen
Athleten wären es 1.2 %. Louvain findet eine ausgeprägte Community-Struktur mit zwei Typen. Elf
grosse Communities bestehen fast nur aus einer Nation. Vier grosse internationale Communities
trennen sich nach Generationen und bilden sich um die Meisterschaften ihrer Zeit, von den
Europameisterschaften der frühen 2000er-Jahre bis zu den aktuellen Weltmeisterschaften.

**2. Zentrale Athleten und Teams** (`05_zentralitaet`). Ein hoher Degree entsteht durch wenige,
sehr grosse internationale Veranstaltungen und nicht durch viele Wettkämpfe. Hohe Betweenness
kennzeichnet Athleten an den Übergängen zwischen Generationen oder zwischen nationaler und
internationaler Ebene. Stärkere Schwimmer sind signifikant zentraler, der Zusammenhang ist mit
ρ zwischen 0.17 und 0.36 aber schwach bis mässig. Nationalteams, deren Athleten sich vor allem
mit Landsleuten treffen, erreichen weniger Athleten anderer Nationen (Spearman −0.82).

**3. Zusammenhalt** (`06_zusammenhalt`). Wiederholte Begegnungen sind kein Zufall. Das
Arbeitsnetz enthält rund zehnmal so viele wiederholte Begegnungen wie das bipartite Configuration
Model. Zusammengehalten wird es von internationalen Meisterschaften. Fehlen die internationalsten
Veranstaltungstage mit 30 % der Teilnahmen, bleiben nur 53.7 % der Athleten in der Giant
Component, bei zufälligen Tagen 91.1 %. Das Entfernen der zentralsten Athleten verändert das Netz
dagegen kaum.

**4. Bevorzugte Verbindungen** (`07_assortativitaet`). Am stärksten verbinden sich Gleichaltrige
(r = 0.85), danach Landsleute (r = 0.42). Starke Schwimmer treffen sich vor allem international
untereinander, in der jüngsten Generation am deutlichsten. National treffen sich dagegen
Gleichaltrige unabhängig von ihrem Niveau.

**5. Vorhersage** (`08_link_prediction`). Netzwerkverfahren sagen neue Begegnungen fünf- bis
sechsmal besser voraus als der Zufall, am besten Soundarajan-Hopcroft mit einer Precision@K von
16.7 %. Die einfache Regel gleiche Nation erreicht aber 25.1 %. Unter den aktiven Athleten trifft
sich mehr als die Hälfte der Landsleute (52.2 %), die sich im Trainingszeitraum nicht begegnet
sind, im Testzeitraum.

## Gesamtbild

Zusammen ergibt sich ein Netz mit zwei Ebenen. National entsteht ein dichtes, wiederkehrendes
Netz unter Gleichaltrigen. International verbinden wenige grosse Meisterschaften die Nationen und
Generationen, und dort treffen sich die stärksten Schwimmer.

## Grenzen

Die Ergebnisse beruhen auf den persönlichen Bestleistungen der TOP100-Listen mit Stand 2025. Pro
Disziplin ist nur die schnellste Zeit der Karriere gespeichert. Verbessert ein Athlet sie,
verschwindet der frühere Tag aus den Daten. Ältere Jahre sind auch deshalb dünner erfasst,
besonders vor 2005. Eine Kante zeigt einen gemeinsamen Bestleistungstag und keine persönliche
Beziehung, und viele tatsächliche Begegnungen an Wettkämpfen fehlen. Zudem betreffen die 126
fehlenden Profile fast nur Frauen, und die Profilnation entspricht dem Stand 2025.

## Ausblick

- **Vollständige Ergebnislisten** statt Bestleistungen würden ein echtes Begegnungsnetz liefern.
  Eine Kante könnte dann auch ein gemeinsames Rennen oder Finale bedeuten.
- **Zeitliche Netze pro Saison** würden zeigen, wie sich Generationen bilden, wie Karrieren
  verlaufen und wie stark die Pandemie 2020 das Netz verändert hat.
- **Netzwerk und Attribute kombiniert** könnten die Link Prediction verbessern, etwa
  Soundarajan-Hopcroft mit Nationen statt Louvain-Communities.
- **Vereine als weitere Relation** würden Trainingsgruppen sichtbar machen, sobald die
  Vereinsangaben vollständiger sind.
- **Diffusion** könnte untersuchen, wie sich Neuerungen wie die Hightech-Anzüge von 2008 und 2009
  über gemeinsame Wettkämpfe ausbreiten. **Graphlets** könnten die Struktur nationaler Netze
  vergleichen.

## Reflexion

Gut funktioniert hat, die Analysen aufeinander aufzubauen. Befunde wie der Kern um die
Weltmeisterschaft 2009 oder die nationale Prägung tauchten in mehreren Notebooks wieder auf und
liessen sich so gegenseitig prüfen.

Anders machen würden wir vor allem die Datengrundlage. Wir haben einen bestehenden Datensatz aus
einem früheren Projekt übernommen, der pro Disziplin nur die schnellste Zeit der Karriere enthält.
Das Netz bildet deshalb gemeinsame Bestleistungstage ab und nicht alle Begegnungen an Wettkämpfen.
Fehlende Profile liessen sich zudem nicht mehr nachladen, da Cloudflare den Zugriff blockierte.
Mit einer eigenen, auf die Fragestellung ausgerichteten Erhebung würden wir vollständige
Ergebnislisten der Wettkämpfe sammeln.

Methodisch haben wir vor allem aus der Projektion gelernt. Jeder Veranstaltungstag erzeugt eine
Clique, weshalb Masse wie Clustering, k-Core oder ein Vergleich mit dem gewöhnlichen Configuration
Model einen passenden Bezugspunkt brauchen. Das bipartite Configuration Model würden wir früher
einsetzen, auch für den Modularity-Test in `04_communities.ipynb`. `07_assortativitaet.ipynb` hat
zudem gezeigt, wie leicht zusammengefasste Gruppen einen Effekt vortäuschen. Ergebnisse auf
Teilgruppen zu prüfen, würden wir von Anfang an einplanen.
