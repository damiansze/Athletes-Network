# 1. Projektübersicht

Dieses Notebook enthält keinen Code. Es beschreibt Thema, Daten und Fragestellungen des Projekts
und zeigt, in welchem Notebook welche Frage untersucht wird. Die Antworten fasst
`09_fazit.ipynb` zusammen.

## Thema

Schwimmer messen sich an nationalen Meisterschaften, an internationalen Wettkampfserien sowie an
Welt- und Europameisterschaften. Ranglisten zeigen, wer wie schnell schwimmt. Sie zeigen aber
nicht, wer wem an Wettkämpfen begegnet und wie sich daraus ein Netz des Schwimmsports bildet.
Genau diesen Blick ermöglicht die soziale Netzwerkanalyse. Wir verbinden Athleten, die am selben
Veranstaltungstag eine Bestleistung erzielten, und untersuchen die Struktur dieses Netzes.

Ziel ist zu verstehen, wie national oder international der Schwimmsport vernetzt ist, welche
Athleten und Veranstaltungen das Netz prägen und ob sich künftige Begegnungen vorhersagen lassen.

## Daten

Grundlage sind die TOP100-Listen von [swimrankings.net](https://www.swimrankings.net) mit bis zu
100 Männern und 100 Frauen je Nation sowie den persönlichen Bestleistungen dieser Athleten,
insgesamt über 13'000 Athleten aus 182 Nationen. Erhoben wurden sie Mitte 2025 mit einem Scraper
aus einem früheren Projekt (`scraper/`).

Pro Athlet ist je Disziplin und Beckenlänge nur die schnellste Zeit der Karriere gespeichert.
Verbessert ein Athlet diese Zeit, ersetzt der neue Eintrag den früheren. Die Daten sind deshalb
keine vollständigen Wettkampfbiografien. Details und Bereinigung stehen in `01_eda.ipynb`.

## Netzwerk

Ein Veranstaltungstag ist die Kombination aus Wettkampfname, Ort, Datum und Beckenlänge. Das
Two-Mode-Netz verbindet Athleten mit den Veranstaltungstagen, an denen sie eine Bestleistung
erzielten. Daraus entsteht die Athletenprojektion. Zwei Athleten sind verbunden, wenn sie am
selben Veranstaltungstag eine ihrer Karrierebestzeiten erzielten, das Kantengewicht zählt die
gemeinsamen Tage. Wenn wir von einer Begegnung sprechen oder davon, dass sich Athleten treffen,
meinen wir einen solchen gemeinsamen Bestleistungstag.

Die meisten Analysen verwenden das Arbeitsnetz mit mindestens zwei gemeinsamen Tagen. Es umfasst
13'156 Athleten und 357'219 Kanten, die Giant Component 12'709 Athleten
(`02_modellierung.ipynb`, `03_reduktion.ipynb`).

## Abgrenzung

Die Untersuchung beschränkt sich auf die gespeicherten Bestleistungen der TOP100-Listen mit Stand
2025. Eine Kante bedeutet, dass zwei Athleten am selben Veranstaltungstag eine ihrer
Karrierebestzeiten erzielten. Sie belegt weder eine persönliche Bekanntschaft noch ein direktes
Rennen gegeneinander. Die Netze fassen alle Jahre zusammen, nur die Link Prediction trennt zwei
Zeiträume. Die verwendeten Verfahren und Masse folgen, wo nicht anders angegeben, dem Skript
*Soziale Netzwerkanalyse* von Michael Henninger (FHNW, 2024).
